# DataCo Supply Chain - Data Cleaning and Preparation

Dataset source: https://www.kaggle.com/datasets/shashwatwork/dataco-smart-supply-chain-for-big-data-analysis

```python
import pandas as pd

df_orig = pd.read_csv("DataCoSupplyChainDataset.csv",encoding="latin1")
df_orig.info()
```

```python
df = df_orig.copy()
```

```python
df.columns.tolist()
```

```python
# Renaming columns based on the official data description

columns_rename = {
    "Type": "transaction_type",

    # Shipping and delivery timing
    "Days for shipping (real)": "actual_shipping_days",
    "Days for shipment (scheduled)": "scheduled_shipping_days",
    "Delivery Status": "delivery_status",
    "Late_delivery_risk": "late_delivery_risk",

    # Financial metrics
    "Benefit per order": "profit_per_order",
    "Sales per customer": "sales_per_customer",

    # Product category
    "Category Id": "category_id",
    "Category Name": "category_name",

    # Customer information
    "Customer Id": "customer_id",
    "Customer Segment": "customer_segment",
    "Customer Fname": "customer_first_name",
    "Customer Lname": "customer_last_name",
    "Customer Email": "customer_email",
    "Customer Password": "customer_password",

    # Store location
    "Customer Street": "store_street",
    "Customer Zipcode": "store_zipcode",
    "Customer City": "store_city",
    "Customer Country": "store_country",
    "Customer State": "store_state",
    "Department Id": "store_department_id",
    "Department Name": "store_department_name",
    "Latitude": "store_latitude",
    "Longitude": "store_longitude",

    # Delivery destination geography
    "Market": "delivery_market",
    "Order Zipcode": "delivery_zipcode",
    "Order City": "delivery_city",
    "Order Country": "delivery_country",
    "Order Region": "delivery_region",
    "Order State": "delivery_state",
   
    # Order information
    "Order Id": "order_id",
    "Order Customer Id": "order_customer_id",
    "order date (DateOrders)": "order_date",
    "Order Status": "order_status",

    # Order item information
    "Order Item Id": "order_item_id",
    "Order Item Cardprod Id": "order_item_product_card_id",
    "Order Item Quantity": "order_item_quantity",
    "Order Item Product Price": "order_item_product_price",
    "Order Item Discount": "order_item_discount",
    "Order Item Discount Rate": "order_item_discount_rate",
    "Order Item Total": "order_item_total",
    "Order Item Profit Ratio": "order_item_profit_ratio",
    "Order Profit Per Order": "order_profit_per_order",

    # Sales
    "Sales": "sales",

    # Product information
    "Product Card Id": "product_card_id",
    "Product Category Id": "product_category_id",
    "Product Name": "product_name",
    "Product Price": "product_price",
    "Product Status": "product_stock_status",
    "Product Description": "product_description",
    "Product Image": "product_link",
    
    # Shipping
    "shipping date (DateOrders)": "shipping_date",
    "Shipping date (DateOrders)": "shipping_date",
    "Shipping Mode": "shipping_mode"
}

df = df.rename(columns=columns_rename)

columns_to_drop = [
    "customer_password",
    "product_link",
    "product_description"
]

df = df.drop(columns=columns_to_drop, errors="ignore")

print(df.shape)
df.head()
```

```python
# Group columns by business area to make the dataset structure clearer.

customer_cols = [
    "customer_id",
    "customer_segment",
    "customer_first_name",
    "customer_last_name",
    "customer_email",
    ]

order_cols = [
    "order_id",
    "order_customer_id",
    "order_date",
    "transaction_type",
    "order_status"
]

delivery_location_cols = [
    "delivery_city",
    "delivery_zipcode",
    "delivery_state",
    "delivery_region",
    "delivery_country",
    "delivery_market",
]

store_cols = [
    "store_department_id",
    "store_department_name",
    "store_street",
    "store_zipcode",
    "store_city",
    "store_state",
    "store_country",
    "store_latitude",
    "store_longitude"
]

product_cols = [
    "product_card_id",
    "product_name",
    "product_category_id",
    "product_price",
    "product_stock_status"
]

category_cols = [
    "category_id",
    "category_name"
]

order_item_cols = [
    "order_item_id",
    "order_item_quantity",
    "order_item_product_price",
    "order_item_discount",
    "order_item_discount_rate",
    "order_item_total",
    "order_item_profit_ratio",
    "order_item_product_card_id"
]

financial_cols = [
    "sales",
    "sales_per_customer",
    "profit_per_order",
    "order_profit_per_order"
]

shipping_cols = [
    "shipping_date",
    "shipping_mode",
    "scheduled_shipping_days",
    "actual_shipping_days",
    "delivery_status",
    "late_delivery_risk"
]

ordered_columns = (
    customer_cols
    + order_cols
    + delivery_location_cols
    + store_cols
    + product_cols
    + category_cols
    + order_item_cols
    + financial_cols
    + shipping_cols
)

# Check whether all expected columns are present and identify any extra columns.
missing_columns = [col for col in ordered_columns if col not in df.columns]
extra_columns = [col for col in df.columns if col not in ordered_columns]

print("Missing columns:", missing_columns)
print("Extra columns:", extra_columns)

df = df[[col for col in ordered_columns if col in df.columns] + extra_columns]

print(df.shape)
df.head()
```

```python
df.to_csv("dataco_clean_columns.csv", index=False)
```

```python
# Check selected columns with missing values or low informational value.
cols_to_check = [
    "customer_email",
    "store_street",
    "delivery_zipcode",
    "product_stock_status"
]

existing_cols_to_check = [col for col in cols_to_check if col in df.columns]

check_cols = pd.DataFrame({
    "missing_values": df[existing_cols_to_check].isna().sum(),
    "unique_values": df[existing_cols_to_check].nunique(dropna=False)
})

check_cols
```

```python
# Drop columns that are not useful for the analysis or final dashboard.
columns_to_drop_2 = [
    "customer_email",
    "store_street",
    "delivery_zipcode",
    "product_stock_status"
   ]

df = df.drop(columns=columns_to_drop_2, errors="ignore")

print(df.shape)
df.head()
```

```python
# Regroup columns after dropping unnecessary fields.
customer_cols = [
    "customer_id",
    "customer_segment",
    "customer_first_name",
    "customer_last_name"
]

delivery_location_cols = [
    "delivery_city",
    "delivery_state",
    "delivery_region",
    "delivery_country",
    "delivery_market"
]

store_cols = [
    "store_department_id",
    "store_department_name",
    "store_zipcode",
    "store_city",
    "store_state",
    "store_country",
    "store_latitude",
    "store_longitude"
]

product_cols = [
    "product_card_id",
    "product_name",
    "product_category_id",
    "product_price"
]

ordered_columns = (
    customer_cols
    + order_cols
    + delivery_location_cols
    + store_cols
    + product_cols
    + category_cols
    + order_item_cols
    + financial_cols
    + shipping_cols
)

missing_columns = [col for col in ordered_columns if col not in df.columns]
extra_columns = [col for col in df.columns if col not in ordered_columns]

print("Missing columns:", missing_columns)
print("Extra columns:", extra_columns)

df = df[[col for col in ordered_columns if col in df.columns] + extra_columns]

print(df.shape)
df.head()
```

```python
df.to_csv("dataco_clean_columns_2.csv", index=False)
```

```python
# Convert date columns to datetime format.
df.info()
```

```python
#Convert date columns to datetime
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["shipping_date"] = pd.to_datetime(df["shipping_date"], errors="coerce")

df[["order_date", "shipping_date"]].info()
```

```python
# Check for duplicated rows.
df.duplicated().sum()
```

```python
# Check dataset granularity: order_item_id should be unique, while order_id can appear multiple times.
print(f"Is order_item_id unique?: {df["order_item_id"].is_unique}")
print(f"How many unique order_item_id are there?: {df['order_item_id'].nunique()}")
print(f"How many order_id are there?: {df["order_id"].nunique()}")
```

```python
# Check consistency between potential primary and foreign keys for the future data model.
print("order_item_product_card_id VS product_card_id:")
print((df['order_item_product_card_id'] == df['product_card_id']).value_counts())
print()
print("order_customer_id VS customer_id:")
print((df['order_customer_id'] == df['customer_id']).value_counts())
print()
print("product_category_id VS category_id:")
print((df['product_category_id'] == df['category_id']).value_counts())

# Some ID columns contain the same values row by row.
# They are kept because they support relationships between the future fact table
# and dimension tables in SQL and Power BI.
```

```python
# Convert store_zipcode from float to nullable integer format.

df["store_zipcode"] = df["store_zipcode"].astype("Int64")

df.to_csv("dataco_clean_columns_3.csv", index=False)
```

```python
# Inspect records with missing store_zipcode values.
df[df["store_zipcode"].isna()][
    [
        "store_city",
        "store_state",
        "store_country",
        "store_latitude",
        "store_longitude"
    ]
]
```

```python
# Check whether the missing zipcodes can be inferred from similar records in the dataset.
df[df["store_zipcode"]==95758][
    [
        "store_city",
        "store_state",
        "store_country",
        "store_latitude",
        "store_longitude",
        "store_zipcode"
    ]
]
```

```python
# Verify that each selected zipcode is consistently associated with a single city.
print("Zipcode: 95758")
print(df[(df["store_zipcode"]==95758) & (df["store_state"]=="CA")]["store_city"].nunique())
print("City Name:")
print(df[(df["store_zipcode"]==95758) & (df["store_state"]=="CA")]["store_city"].unique())
print()
print("Zipcode: 91732")
print(df[(df["store_zipcode"]==91732) & (df["store_state"]=="CA")]["store_city"].nunique())
print("City Name:")
print(df[(df["store_zipcode"]==91732) & (df["store_state"]=="CA")]["store_city"].unique())
```

```python
# Manual correction of shifted store location fields

df.loc[[35704, 46440], "store_city"] = "Elk Grove"
df.loc[[35704, 46440], "store_state"] = "CA"
df.loc[[35704, 46440], "store_zipcode"] = 95758

df.loc[82511, "store_city"] = "El Monte"   
df.loc[82511, "store_state"] = "CA"
df.loc[82511, "store_zipcode"] = 91732

df.loc[
    [35704, 46440, 82511],
    ["store_city", "store_state", "store_country", "store_zipcode", "store_latitude", "store_longitude"]
]
```

```python
# Three records had shifted store location values:
# store_city contained the state, store_state contained the zipcode, and store_zipcode was missing.
# They were manually corrected using matching zipcode-city-state combinations already present in the dataset.

df["store_zipcode"].isna().sum()
```

```python
# Standardize the store_country label.
print(f"Before: {df["store_country"].value_counts(dropna=False)}")

df["store_country"] = df["store_country"].replace({
    "EE. UU.": "USA"
})
print()
print(f"After: {df["store_country"].value_counts(dropna=False)}")
```

```python
# Remove leading and trailing spaces from text columns only

text_cols = df.select_dtypes(include="object").columns

for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
# Create a final data quality summary to review data types, missing values, and unique values.
data_quality_summary = pd.DataFrame({
    "dtype": df.dtypes,
    "missing_values": df.isna().sum(),
    "missing_pct": df.isna().mean() * 100,
    "unique_values": df.nunique()
})

data_quality_summary.sort_values("missing_pct", ascending=False).head(20)
```

```python
# Save the final cleaned dataset for SQL import and Power BI modeling.
df_clean = df.copy()
df_clean.to_csv("dataco_cleaned_final.csv", index=False)
```
